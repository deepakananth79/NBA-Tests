```python
import pandas as pd

# Show all columns in Jupyter output
pd.set_option('display.max_columns', None)
```

```python
from nba_api.stats.static import players, teams
from nba_api.stats.endpoints import playercareerstats

# 1. Get the DataFrame
career = playercareerstats.PlayerCareerStats(player_id='203999') 
df = career.get_data_frames()[0]  # ← this line defines df

# 2. Ensure ID columns are integers
df['PLAYER_ID'] = df['PLAYER_ID'].astype(int)
df['TEAM_ID'] = df['TEAM_ID'].astype(int)

# 3. Create team lookup and map team names
team_lookup = {t['id']: t['full_name'] for t in teams.get_teams()}
df['TEAM_NAME'] = df['TEAM_ID'].map(team_lookup)
```

```python
# 4. Build a player name lookup using commonplayerinfo
from nba_api.stats.endpoints import commonplayerinfo
import time

player_lookup = {}
for pid in df['PLAYER_ID'].unique():
    try:
        info = commonplayerinfo.CommonPlayerInfo(player_id=pid)
        name = info.get_data_frames()[0].loc[0, 'DISPLAY_FIRST_LAST']
        player_lookup[pid] = name
        time.sleep(0.6)
    except:
        player_lookup[pid] = "Unknown"

df['PLAYER_NAME'] = df['PLAYER_ID'].map(player_lookup)
```

```python
cols = list(df.columns)

# Remove PLAYER_NAME and TEAM_NAME from the current list
cols.remove('PLAYER_NAME')
cols.remove('TEAM_NAME')
cols.remove('PLAYER_ID')
cols.remove('TEAM_ID')
cols.remove('LEAGUE_ID')

# Insert PLAYER_NAME as 2nd column (index 1)
cols.insert(1, 'PLAYER_NAME')

# Insert TEAM_NAME as 3rd column (index 2)
cols.insert(2, 'TEAM_NAME')

# Reorder the DataFrame columns
df = df[cols]
```
